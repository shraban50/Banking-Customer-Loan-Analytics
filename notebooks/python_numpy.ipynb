## 11. Missing Value Profiling

Before cleaning the datasets, we need to understand whether missing
values exist and where they occur.

The following function calculates:
- Missing count
- Missing percentage

```python
def missing_summary(df):
    result = pd.DataFrame({
        "Missing": df.isna().sum(),
        "Missing %": (df.isna().mean() * 100).round(2)
    })
    
    return result.sort_values("Missing", ascending=False)
```

## 33. Exception Handling

Exception handling is used to prevent a program from stopping

In banking data analysis, exception handling can be useful when:

- Reading datasets

Python uses try, except, else and finally blocks for exception handling.

```python
try:
    balance = accounts["Balance"].iloc[0]
    print("Account Balance:", balance)

except Exception as error:
    print("An error occurred:", error)
```

### 34 Handling a Value Error

A ValueError can occur when a value cannot be converted
to the required data type.

```python
try:
    loan_amount = float("50000")
    print("Loan Amount:", loan_amount)

except ValueError:
    print("Invalid loan amount.")
```

### 34.2 Handling Invalid Numeric Data

The following example demonstrates how invalid numeric
values can be handled safely.

```python
try:
    amount = float("invalid_amount")
    print("Transaction Amount:", amount)

except ValueError:
    print("Transaction amount is not a valid number.")
```

### 34.3 Handling Missing Dataset Columns

Exception handling can also be used when accessing
a column that may not exist.

```python
try:
    income_values = customers["Income"]
    print("Income column accessed successfully.")

except KeyError:
    print("Income column does not exist in the dataset.")
```

### 34.4 Using Multiple Exceptions

```python
try:
    account_id = accounts["AccountID"].iloc[0]
    balance = float(accounts["Balance"].iloc[0])

    print("Account ID:", account_id)
    print("Balance:", balance)

except KeyError:
    print("Required column is missing.")

except IndexError:
    print("No account record is available.")

except ValueError:
    print("Balance contains an invalid numeric value.")
```

### 34.5 Using Finally

The finally block executes whether an exception occurs or not.

```python
try:
    transaction_amount = float(transactions["Amount"].iloc[0])
    print("Transaction Amount:", transaction_amount)

except (KeyError, IndexError, ValueError) as error:
    print("Error:", error)

finally:
    print("Transaction check completed.")
```

### 34.6 Reusable Function with Exception Handling

A reusable function can combine functions and exception handling
to safely calculate loan exposure.

```python
def calculate_loan_exposure(loan_amount, repayment_amount):
    try:
        loan_amount = float(loan_amount)
        repayment_amount = float(repayment_amount)

        outstanding = loan_amount - repayment_amount

        if outstanding < 0:
            outstanding = 0

        return outstanding

    except (TypeError, ValueError):
        return None
```

```python
print(
    "Outstanding Amount:",
    calculate_loan_exposure(100000, 40000)
)

print(
    "Invalid Input:",
    calculate_loan_exposure("abc", 40000)
)
```
